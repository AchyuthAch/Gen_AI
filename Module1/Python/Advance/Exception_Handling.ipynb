## Exception Handling

### Run time Errors

```python
a = int(input("Enter the value of a: "))
b = int(input("Enter the value of b: "))
print(a/b)
```

```python
try:
    a = int(input("Enter the value of a: "))
    b = int(input("Enter the value of b: "))
    print(a/b)
except Exception as e:
    print(e)
```
