### Basic of using Jupyter Notebook

- Rules

    - User Enter to add new line in same block <br/>
    - Use shift+enter to enter into new block  <br/>
    - Use ctrl+enter to run the same block    <br/>
    - Use ESC+A to enter new block above <br/>
    - Use ESC+B to enter new block below<br/>
    - User ESC+DD to delete the current block<br/>
