```python
## IF & IFELSE
x = 1
```

```python
if (x >2):
    print(f"Yes {x} is greater than 2")
    print("Complete")
else:
    print(f" No {x} is not greater than 2")
```

```python
## While
```

```python
x = 1
```

```python
while(x <=10):
    print(x)

    if x == 5:
        break
    x = x+1
print("Loop Ended")
```
